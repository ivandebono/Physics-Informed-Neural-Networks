# Session 3: Introduction to neural networks and deep learning 

We are now transitioning from classical numerical methods for partial differential equations (PDEs) to the world of neural networks, which are the foundation of physics-informed neural networks (PINNs). As a physicist, I find neural networks fascinating because they act like universal function approximators, much like Fourier series or Taylor expansions, but with the power to handle highly non-linear, high-dimensional problems that plague traditional grids.

This session focuses on the basics: what neural networks are, why they can approximate functions, activation functions, and the all-important backpropagation algorithm.

## 1. Feed-forward neural networks (multilayer perceptrons)

A basic neural network is a stack of layers of "neurons". Each neuron computes a weighted sum of inputs plus a bias, then applies a non-linear **activation function**.

A typical feed-forward network has:
- Input layer
- Hidden layers (usually several)
- Output layer

Mathematically, for a network with $L$ layers:
$$
\mathbf{h}^{(l+1)} = \sigma(\mathbf{W}^{(l)} \mathbf{h}^{(l)} + \mathbf{b}^{(l)})
$$
where $\sigma$ is the activation, $\mathbf{W}, \mathbf{b}$ are learnable weights and biases.

#### 2. The universal approximation theorem
This is the key theoretical justification for using neural networks in scientific computing.

**Theorem** (Cybenko 1989; Hornik 1991): A feed-forward neural network with a single hidden layer containing a finite number of neurons with a non-linear activation (e.g., sigmoid) can approximate any continuous function on a compact domain to arbitrary accuracy.

In practice, deeper networks are more efficient. See https://www.deep-mind.org/2023/03/26/the-universal-approximation-theorem for an excellent explanation.








This is why, in PINNs, we can use a neural network to represent the unknown solution $u(x,t)$ of a PDE. It is guaranteed (in theory) to be able to represent it.

## 3. Activation functions

Linearity alone would make the whole network equivalent to a single linear transformation, which is useless for complex physics. Non-linear activations are essential.

Common ones:












- **Sigmoid**: $\sigma(z) = \frac{1}{1+e^{-z}}$ → outputs [0,1], but suffers from vanishing gradients.
- **Tanh**: $\sigma(z) = \tanh(z)$ → outputs [-1,1], zero-centred, still vanishing gradients.
- **Rectified Linear Unit (ReLU)**: $\sigma(z) = \max(0,z)$ → fast, avoids vanishing gradients, but "dying ReLU" problem.
- In PINNs, **tanh** is very common because it is smooth and has good higher-order derivatives (crucial for PDE residuals).

#### 4. Training: backpropagation and gradient descent

We train by minimising a loss function $\mathcal{L}(\theta)$ (e.g., mean squared error) w.r.t. parameters $\theta = \{\mathbf{W}, \mathbf{b}\}$.

**Backpropagation** efficiently computes gradients using the chain rule, propagating errors backwards through the network.








Then update: $\theta \leftarrow \theta - \eta \nabla_\theta \mathcal{L}$ (gradient descent), or better with Adam, etc.
