# Introduction

As a physicist with extensive experience in computational methods and
machine learning, having worked in academia (astrophysics and cosmology)
and industry (AI, NLPs, LLMs and all that), I am delighted present this
tutorial course on Physics-Informed Neural Networks (PINNs). This topic
bridges classical physics, governed by differential equations, and
modern deep learning, enabling us to solve complex physical problems
efficiently, even with limited data.

PINNs are particularly powerful for scientific computing, as they embed
physical laws directly into neural network training, ensuring solutions
respect conservation laws, symmetries, and other principles from
physics.

This course is pitched at an undergraduate-level audience, with
backgrounds in physics, applied mathematics, or engineering. It assumes
familiarity with partial differential equations (PDEs), basic linear
algebra, calculus, and introductory programming. I will use Python, but
anyone with knowledge of programming in other languages should be able
to understand the code. No prior deep learning experience is required.
The point of this tutorial is to build this from scratch.

## Course objectives

By the end, you will:

-   Understand how neural networks can approximate solutions to
    differential equations.

-   Implement PINNs to solve forward and inverse problems in physics.

-   Critically evaluate the strengths and limitations of PINNs compared
    to traditional numerical methods (e.g., finite differences or finite
    elements).

-   Apply PINNs to real physical systems, such as heat transfer, wave
    propagation, and fluid dynamics.

## Syllabus

The syllabus should take up about 36 hours of your time, including
reading this material, tinkering with the code, and doing some
background reading.

-   **Sessions 1--2: Foundations of differential equations in physics**

    -   Review of ordinary differential equations (ODEs) and PDEs common
        in physics (e.g., heat equation, wave equation, and Burgers'
        equation).

    -   Classical numerical solvers: finite differences and
        explicit/implicit schemes.

    -   Limitations of traditional methods: mesh generation,
        computational cost, and the curse of dimensionality.

-   **Sessions 3--4: Introduction to neural networks and deep learning**

    -   Universal approximation theorem: why neural networks can
        represent any continuous function.

    -   Feed-forward networks and activation functions such as ReLU,
        tanh, and sigmoid.

    -   Backpropagation and gradient descent.

    -   Automatic differentiation, which is key to PINNs: how libraries
        like PyTorch and TensorFlow compute exact derivatives
        efficiently.

    -   Hands-on exercise: simple function approximation with a neural
        network in Python.

-   **Sessions 5--6: Core concept of PINNs**

    -   The original PINN framework (Raissi et al., 2019): embedding PDE
        residuals into the loss function.

    -   Loss components: data fitting, PDE residual, and
        boundary/initial conditions.

    -   Forward problems: solving PDEs with known parameters.

    -   Inverse problems: inferring unknown parameters (e.g., material
        properties) from data.

    -   Example: solving the 1D heat equation.

    -   Mathematical formulation: consider a PDE
        $\mathcal{N}[u] = f(x, t)$, where $\mathcal{N}$ is a
        differential operator.

    -   The PINN approximates $u(x, t; \theta)$ with a neural network
        parameterized by $\theta$.

    -   The loss is
        $$\mathcal{L} = \mathcal{L}_{\text{data}} + \lambda_r \mathcal{L}_{\text{residual}} + \lambda_b \mathcal{L}_{\text{boundary}},$$
        where
        $$\mathcal{L}_{\text{residual}} = \frac{1}{N_r} \sum \lvert \mathcal{N}[u(x_i, t_i; \theta)] - f(x_i, t_i) \rvert^2.$$

-   **Sessions 7--8: Implementation in Python**

    -   PyTorch for PINNs (my preferred framework for its flexibility
        and physics community adoption).

    -   Sampling strategies: collocation points (Latin hypercube and
        uniform grid).

    -   Basic code structure: define the network, compute residuals via
        autograd, and train with the Adam optimizer.

    -   Hands-on coding:

        -   Burgers' equation (viscous shock propagation). A minimal
            PyTorch example for the 1D Burgers' equation
            ($u_t + u u_x = \nu u_{xx}$) is found in
            `burgers_1d_eqn.py`.

        -   Poisson equation in 2D (electrostatics).

    -   Debugging tips: vanishing gradients and loss balancing.

-   **Sessions 9--10: Advanced topics and extensions**

    -   Handling stiff equations and failures, such as spectral bias at
        high frequencies.

    -   Adaptive sampling, weighted losses, and domain decomposition.

    -   Conservative PINNs (cPINNs) and variational formulations.

    -   Recent advances: hybrid PINN--Fourier Neural Operators,
        multi-fidelity PINNs, and Physics-Informed Kolmogorov--Arnold
        Networks (PIKANs) for improved expressivity.

    -   Uncertainty quantification and inverse problems with noisy data.

-   **Sessions 11--12: Applications and projects**

    -   Physics examples: Schrödinger equation (quantum mechanics),
        Navier--Stokes equations (fluid dynamics), and
        reaction--diffusion systems (pattern formation).

    -   Project: choose a physical system, such as a pendulum with
        unknown damping or heat flow in an irregular domain, and
        implement a PINN solver.

    -   Comparison with traditional numerical methods.
